```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset loaded successfully!")
```

```python
print("First 10 Rows:")
display(df.head(10))
```

```python
print("Last 5 Rows:")
display(df.tail(5))
```

```python
print("Dataset Dimensions:")
print(df.shape)
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Summary Statistics:")
display(df.describe())

print("Age Mean:", df["Age"].mean())
print("Age Minimum:", df["Age"].min())
print("Age Maximum:", df["Age"].max())

print("Fare Mean:", df["Fare"].mean())
print("Fare Minimum:", df["Fare"].min())
print("Fare Maximum:", df["Fare"].max())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
median_age = df["Age"].median()

df["Age"] = df["Age"].fillna(median_age)

print("Missing Age values after filling:")
print(df["Age"].isnull().sum())
```

```python
survived = df[df["Survived"] == 1]

print("Surviving Passengers:")
display(survived)
```

```python
fare_array = df["Fare"].to_numpy()

print("Fare as NumPy Array:")
print(fare_array)
```

```python
maximum_fare = np.max(fare_array)
average_fare = np.mean(fare_array)

print("Maximum Fare:", maximum_fare)
print("Average Fare:", average_fare)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=20, edgecolor="black")

plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
survival_count = df["Survived"].value_counts().sort_index()

plt.figure(figsize=(7, 5))

plt.bar(["Not Survived", "Survived"], survival_count)

plt.title("Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Age"], df["Fare"], alpha=0.6)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```
